# M5 Dataset Check
This notebook verifies that the 3 raw M5 CSV files load correctly via PySpark.

```python
import os
os.environ["PYSPARK_PYTHON"] = r"D:\Retail Demand Forecasting\.venv\Scripts\python.exe"
os.environ["PYSPARK_DRIVER_PYTHON"] = r"D:\Retail Demand Forecasting\.venv\Scripts\python.exe"

from pyspark.sql import SparkSession

PROJECT_ROOT = r"D:\Retail Demand Forecasting"

spark = (
    SparkSession.builder
    .master("local[*]")
    .appName("m5_dataset_check")
    .config("spark.sql.shuffle.partitions", "4")
    .config("spark.driver.memory", "4g")
    .getOrCreate()
)
print(f"Spark version: {spark.version}")
```

## 1. Calendar

```python
calendar = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(os.path.join(PROJECT_ROOT, "data/01_raw/calendar.csv"))
)
print(f"calendar: {calendar.count():,} rows x {len(calendar.columns)} columns")
calendar.limit(10).toPandas()
```

## 2. Sell Prices

```python
sell_prices = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(os.path.join(PROJECT_ROOT, "data/01_raw/sell_prices.csv"))
)
print(f"sell_prices: {sell_prices.count():,} rows x {len(sell_prices.columns)} columns")
sell_prices.limit(10).toPandas()
```

## 3. Sales Train Validation

```python
sales_train = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(os.path.join(PROJECT_ROOT, "data/01_raw/sales_train_validation.csv"))
)
print(f"sales_train: {sales_train.count():,} rows x {len(sales_train.columns)} columns")
sales_train.limit(10).toPandas()
```

## 4. Schema Summary

```python
for name, df in [("calendar", calendar), ("sell_prices", sell_prices), ("sales_train", sales_train)]:
    print(f"\n=== {name} ===")
    print(f"  Rows: {df.count():,}")
    print(f"  Columns: {len(df.columns)}")
    print(f"  First 10 columns: {df.columns[:10]}")
    if name == "sales_train":
        print(f"  (has {len(df.columns) - 6} day columns: d_1 ... d_{len(df.columns) - 6})")
```

```python
spark.stop()
print("Spark session stopped.")
```
