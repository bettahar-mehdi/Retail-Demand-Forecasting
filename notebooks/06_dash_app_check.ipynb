# Dash Frontend — Launch & Verify
Start the Dash app and the FastAPI backend, then verify both work together.

```python
import os
import subprocess, time, sys
os.chdir(r"D:\Retail Demand Forecasting")
```

## 1. Start FastAPI backend

```python
api_proc = subprocess.Popen(
    [sys.executable, "-m", "uvicorn", "retail_demand_forecasting.api.app:app",
     "--host", "127.0.0.1", "--port", "8000"],
    cwd=r"D:\Retail Demand Forecasting",
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL,
)
time.sleep(12)

import requests
r = requests.get("http://127.0.0.1:8000/health", timeout=5)
print(f"API health: {r.json()}")
```

## 2. Start Dash frontend

```python
dash_proc = subprocess.Popen(
    [sys.executable, "-m", "retail_demand_forecasting.dash_app.app"],
    cwd=r"D:\Retail Demand Forecasting",
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL,
)
time.sleep(5)
print("Dash running at http://localhost:8050")
print("API running at http://localhost:8000")
```

## 3. Verify API from Dash

```python
r = requests.post("http://127.0.0.1:8000/predict", json={
    "lag_7": 5, "lag_28": 3, "rolling_mean_7": 4.5, "rolling_mean_28": 3.2,
    "day_of_week": 6, "month": 4, "year": 2016,
    "snap_CA": 1, "snap_TX": 0, "snap_WI": 0,
    "has_event_1": 0, "has_event_2": 0, "sell_price": 1.25,
}, timeout=10)
print(f"Prediction: {r.json()}")
```

```python
print("\n=== Stopping servers ===")
api_proc.terminate()
dash_proc.terminate()
print("Done.")
```
