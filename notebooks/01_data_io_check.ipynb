# Phase 1 — Data I/O Check
Load raw M5 datasets via Kedro catalog and verify they are accessible.

```python
import os
os.environ["PYSPARK_PYTHON"] = r"D:\Retail Demand Forecasting\.venv\Scripts\python.exe"
os.environ["PYSPARK_DRIVER_PYTHON"] = r"D:\Retail Demand Forecasting\.venv\Scripts\python.exe"

os.chdir(r"D:\Retail Demand Forecasting")
```

```python
from kedro.framework.session import KedroSession
from kedro.framework.startup import bootstrap_project

bootstrap_project(r"D:\Retail Demand Forecasting")
session = KedroSession.create(project_path=r"D:\Retail Demand Forecasting")
context = session.load_context()
catalog = context.catalog
print("Catalog datasets:", list(catalog.keys()))
```

## 1. Load calendar_raw

```python
calendar_raw = catalog.load("calendar_raw")
print(f"calendar_raw: {calendar_raw.count():,} rows x {len(calendar_raw.columns)} columns")
calendar_raw.limit(5).toPandas()
```

## 2. Load sell_prices_raw

```python
sell_prices_raw = catalog.load("sell_prices_raw")
print(f"sell_prices_raw: {sell_prices_raw.count():,} rows x {len(sell_prices_raw.columns)} columns")
sell_prices_raw.limit(5).toPandas()
```

## 3. Load sales_train_raw

```python
sales_train_raw = catalog.load("sales_train_raw")
print(f"sales_train_raw: {sales_train_raw.count():,} rows x {len(sales_train_raw.columns)} columns")
sales_train_raw.limit(5).toPandas()
```
